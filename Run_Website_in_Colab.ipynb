## 🌐 Step 5: Run the Server with Public URL

**IMPORTANT**: Run this cell and wait for the public URL to appear. Click the URL to view the website!

```python
import subprocess
import threading
import time
import re
from IPython.display import display, HTML

# Function to run the dev server
def run_dev_server():
    subprocess.run(['npm', 'run', 'dev'], cwd='/content/cloud_test1')

# Function to run localtunnel
def run_tunnel():
    time.sleep(10)  # Wait for dev server to start
    result = subprocess.run(
        ['lt', '--port', '5173', '--subdomain', f'plant-hub-{int(time.time())}'],
        capture_output=True,
        text=True
    )
    print(result.stdout)
    print(result.stderr)

# Start the dev server in a background thread
server_thread = threading.Thread(target=run_dev_server, daemon=True)
server_thread.start()

print("🔄 Starting development server...")
print("⏳ Waiting for server to be ready...\n")

# Wait a bit for the server to start
time.sleep(15)

# Start tunnel in background
tunnel_thread = threading.Thread(target=run_tunnel, daemon=True)
tunnel_thread.start()

time.sleep(5)

display(HTML("""
<div style="background: linear-gradient(135deg, #667eea 0%, #764ba2 100%); 
            padding: 30px; 
            border-radius: 15px; 
            color: white;
            text-align: center;
            margin: 20px 0;
            box-shadow: 0 10px 40px rgba(0,0,0,0.3);">
    <h2 style="margin: 0 0 15px 0;">✅ Server is Running!</h2>
    <p style="font-size: 18px; margin: 10px 0;">The React website is now live!</p>
    <p style="font-size: 14px; opacity: 0.9;">Check the output above for the public URL (starts with https://)</p>
    <p style="font-size: 14px; opacity: 0.9; margin-top: 15px;">Or use the local URL: <strong>http://localhost:5173</strong></p>
</div>
"""))

print("\n" + "="*60)
print("🌐 Access the website using one of the URLs above")
print("💡 The server will keep running as long as this cell is active")
print("🛑 To stop: Interrupt the kernel (Runtime → Interrupt)")
print("="*60)

# Keep the cell running
try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("\n🛑 Server stopped")
```

## 🎯 Alternative: Using ngrok (More Reliable)

If localtunnel doesn't work, use ngrok instead. This provides a more stable tunnel.

```python
# Install pyngrok
!pip install pyngrok -q

from pyngrok import ngrok
import subprocess
import threading
import time
from IPython.display import display, HTML

# Kill any existing ngrok tunnels
ngrok.kill()

# Function to run the dev server
def run_dev_server():
    subprocess.run(['npm', 'run', 'dev'], cwd='/content/cloud_test1')

# Start the dev server in a background thread
server_thread = threading.Thread(target=run_dev_server, daemon=True)
server_thread.start()

print("🔄 Starting development server...")
time.sleep(15)  # Wait for server to start

# Create ngrok tunnel
public_url = ngrok.connect(5173)

display(HTML(f"""
<div style="background: linear-gradient(135deg, #a855f7 0%, #06b6d4 100%); 
            padding: 40px; 
            border-radius: 20px; 
            color: white;
            text-align: center;
            margin: 30px 0;
            box-shadow: 0 20px 60px rgba(168, 85, 247, 0.4);">
    <h1 style="margin: 0 0 20px 0; font-size: 32px;">🎉 Plant Intelligence Hub is Live!</h1>
    <div style="background: rgba(255,255,255,0.2);
                padding: 20px;
                border-radius: 10px;
                margin: 20px 0;">
        <p style="font-size: 16px; margin: 5px 0;">Public URL:</p>
        <a href="{public_url}" target="_blank" 
           style="color: white; 
                  font-size: 24px; 
                  font-weight: bold;
                  text-decoration: none;
                  display: inline-block;
                  padding: 15px 30px;
                  background: rgba(255,255,255,0.3);
                  border-radius: 10px;
                  margin-top: 10px;
                  transition: all 0.3s;">
            🌐 {public_url}
        </a>
    </div>
    <p style="font-size: 16px; opacity: 0.9; margin-top: 20px;">
        ✨ Click the link above to view the actual React website!
    </p>
    <p style="font-size: 14px; opacity: 0.8; margin-top: 10px;">
        🔄 The site is running with hot-reload - changes will update automatically
    </p>
</div>
"""))

print("\n" + "="*60)
print(f"🌐 Public URL: {public_url}")
print("💡 Server is running with hot-reload enabled")
print("🛑 To stop: Interrupt the kernel")
print("="*60)

# Keep the cell running
try:
    while True:
        time.sleep(1)
except KeyboardInterrupt:
    print("\n🛑 Server stopped")
    ngrok.kill()
```

## 📊 Server Status

Check if the server is running properly.

```python
import requests

try:
    response = requests.get('http://localhost:5173', timeout=5)
    if response.status_code == 200:
        print("✅ Server is running properly!")
        print(f"Status Code: {response.status_code}")
    else:
        print(f"⚠️ Server returned status code: {response.status_code}")
except Exception as e:
    print(f"❌ Server is not responding: {str(e)}")
    print("\n💡 Make sure you've run the server cell above first!")
```
