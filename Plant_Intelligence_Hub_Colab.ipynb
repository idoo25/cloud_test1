## Step 2: Import Libraries

```python
import gradio as gr
import plotly.graph_objects as go
from plotly.subplots import make_subplots
import numpy as np
import pandas as pd
from datetime import datetime, timedelta
from PIL import Image
import random
import io
```

## Step 3: Generate Mock Sensor Data

```python
def generate_sensor_data():
    """Generate current sensor readings"""
    return {
        'temperature': {'value': round(20 + random.uniform(0, 10), 1), 'unit': '°C', 'status': 'good', 'min': 18, 'max': 30},
        'humidity': {'value': round(50 + random.uniform(0, 30), 1), 'unit': '%', 'status': 'good', 'min': 50, 'max': 80},
        'soil_moisture': {'value': round(30 + random.uniform(0, 30), 1), 'unit': '%', 'status': 'warning', 'min': 30, 'max': 60},
        'light_intensity': {'value': int(200 + random.uniform(0, 800)), 'unit': 'lux', 'status': 'good', 'min': 600, 'max': 1000}
    }

def generate_historical_data(days=7):
    """Generate historical sensor data for plotting"""
    hours = days * 24
    timestamps = [datetime.now() - timedelta(hours=i) for i in range(hours, 0, -1)]
    
    data = {
        'timestamp': timestamps,
        'temperature': [20 + random.uniform(0, 10) for _ in range(hours)],
        'humidity': [50 + random.uniform(0, 30) for _ in range(hours)],
        'soil_moisture': [30 + random.uniform(0, 30) for _ in range(hours)],
        'light_intensity': [200 + random.uniform(0, 800) for _ in range(hours)]
    }
    
    return pd.DataFrame(data)
```

## Step 4: AI Analysis Functions

```python
def analyze_plant_image(image):
    """Analyze uploaded plant image and return health metrics"""
    if image is None:
        return "No image uploaded. Please upload a plant image for analysis."
    
    # Simulate AI analysis
    health_score = random.randint(75, 95)
    
    analysis_result = f"""
    ### 🌿 AI Analysis Results
    
    **Health Score:** {health_score}% ✅
    
    #### 🔬 Disease Detection
    - Status: **Healthy**
    - Severity: Good
    - Details: No diseases detected
    
    #### 💧 Water Stress Level
    - Status: **Low**
    - Severity: Warning
    - Details: Slightly below optimal
    
    #### 🐛 Pest Identification
    - Status: **Clear**
    - Severity: Good
    - Details: No pests identified
    
    #### ✅ Care Recommendations
    1. Monitor soil moisture levels
    2. Ensure adequate sunlight (6-8 hours)
    3. Check for nutrient deficiency
    
    **Overall Condition:** Excellent condition detected. Continue monitoring.
    """
    
    return analysis_result

def get_care_recommendations():
    """Get AI-powered care recommendations"""
    recommendations = [
        "💧 Water your plant - Soil moisture is slightly low",
        "☀️ Increase light exposure - Move to brighter location",
        "🌡️ Temperature is optimal - Maintain current conditions",
        "🌱 Consider fertilizing - It's been 2 weeks since last feeding"
    ]
    return "\n".join(recommendations)
```

## Step 5: Data Visualization Functions

```python
def create_sensor_display():
    """Create formatted sensor data display"""
    sensors = generate_sensor_data()
    
    display_text = "## 📊 Current Sensor Readings\n\n"
    
    icons = {
        'temperature': '🌡️',
        'humidity': '💧',
        'soil_moisture': '🌱',
        'light_intensity': '☀️'
    }
    
    labels = {
        'temperature': 'Temperature',
        'humidity': 'Humidity',
        'soil_moisture': 'Soil Moisture',
        'light_intensity': 'Light Intensity'
    }
    
    for key, data in sensors.items():
        status_icon = '✅' if data['status'] == 'good' else '⚠️'
        display_text += f"### {icons[key]} {labels[key]}\n"
        display_text += f"**Value:** {data['value']} {data['unit']} {status_icon}\n"
        display_text += f"**Range:** {data['min']}-{data['max']} {data['unit']}\n\n"
    
    return display_text

def create_historical_chart(period='week'):
    """Create interactive historical data chart"""
    days_map = {'week': 7, 'month': 30, 'quarter': 90}
    days = days_map.get(period, 7)
    
    df = generate_historical_data(days)
    
    # Create subplots
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=('Temperature (°C)', 'Humidity (%)', 'Soil Moisture (%)', 'Light Intensity (lux)'),
        vertical_spacing=0.12,
        horizontal_spacing=0.1
    )
    
    # Temperature
    fig.add_trace(
        go.Scatter(x=df['timestamp'], y=df['temperature'], name='Temperature', 
                   line=dict(color='#ef4444', width=2)),
        row=1, col=1
    )
    
    # Humidity
    fig.add_trace(
        go.Scatter(x=df['timestamp'], y=df['humidity'], name='Humidity',
                   line=dict(color='#3b82f6', width=2)),
        row=1, col=2
    )
    
    # Soil Moisture
    fig.add_trace(
        go.Scatter(x=df['timestamp'], y=df['soil_moisture'], name='Soil Moisture',
                   line=dict(color='#06b6d4', width=2)),
        row=2, col=1
    )
    
    # Light Intensity
    fig.add_trace(
        go.Scatter(x=df['timestamp'], y=df['light_intensity'], name='Light Intensity',
                   line=dict(color='#f59e0b', width=2)),
        row=2, col=2
    )
    
    # Update layout
    fig.update_layout(
        height=700,
        showlegend=False,
        title_text=f"📈 Historical Trends - Last {period.capitalize()}",
        template='plotly_dark',
        paper_bgcolor='#0a0a0f',
        plot_bgcolor='#1a1a2e'
    )
    
    fig.update_xaxes(showgrid=True, gridwidth=1, gridcolor='#2d2d3f')
    fig.update_yaxes(showgrid=True, gridwidth=1, gridcolor='#2d2d3f')
    
    return fig
```

## Step 7: Create Main Gradio Interface

```python
# Custom CSS for dark theme similar to original
custom_css = """
.gradio-container {
    background: linear-gradient(135deg, #0a0a0f 0%, #1a1a2e 100%);
    font-family: 'Inter', sans-serif;
}

.contain {
    max-width: 1600px;
    margin: 0 auto;
}

button {
    background: linear-gradient(135deg, #a855f7 0%, #06b6d4 100%) !important;
    border: none !important;
    border-radius: 12px !important;
    font-weight: 600 !important;
}

.panel {
    background: rgba(26, 26, 46, 0.5) !important;
    backdrop-filter: blur(12px) !important;
    border: 1px solid rgba(168, 85, 247, 0.2) !important;
    border-radius: 16px !important;
}
"""

# Create the Gradio interface
with gr.Blocks(css=custom_css, theme=gr.themes.Soft(primary_hue="purple", secondary_hue="cyan")) as demo:
    
    # Header
    gr.Markdown("""
    # 🌿 Plant Intelligence Hub
    ### Real-time monitoring powered by AI technology
    """)
    
    # Main tabs
    with gr.Tabs():
        
        # Tab 1: Dashboard Overview
        with gr.Tab("📊 Dashboard"):
            with gr.Row():
                with gr.Column(scale=1):
                    sensor_display = gr.Markdown(value=create_sensor_display())
                    refresh_btn = gr.Button("🔄 Refresh Sensors", size="sm")
                    refresh_btn.click(fn=create_sensor_display, outputs=sensor_display)
                
                with gr.Column(scale=1):
                    gr.Markdown("## 💡 Quick Recommendations")
                    recommendations = gr.Markdown(value=get_care_recommendations())
        
        # Tab 2: Plant Analysis
        with gr.Tab("🔬 AI Analysis"):
            with gr.Row():
                with gr.Column(scale=1):
                    gr.Markdown("## 📸 Upload Plant Image")
                    image_input = gr.Image(type="pil", label="Drop your image here or click to upload")
                    analyze_btn = gr.Button("🤖 Analyze Plant", variant="primary")
                
                with gr.Column(scale=1):
                    analysis_output = gr.Markdown(label="Analysis Results")
            
            analyze_btn.click(fn=analyze_plant_image, inputs=image_input, outputs=analysis_output)
            
            # Example images
            gr.Markdown("### 📷 Example Images")
            gr.Examples(
                examples=[
                    ["https://images.unsplash.com/photo-1466692476868-aef1dfb1e735?w=400"],
                    ["https://images.unsplash.com/photo-1530836369250-ef72a3f5cda8?w=400"],
                ],
                inputs=image_input
            )
        
        # Tab 3: Historical Data
        with gr.Tab("📈 Historical Data"):
            gr.Markdown("## 📊 Sensor Trends Over Time")
            
            period_selector = gr.Radio(
                choices=['week', 'month', 'quarter'],
                value='week',
                label="Select Time Period"
            )
            
            chart_output = gr.Plot(value=create_historical_chart('week'))
            
            period_selector.change(
                fn=create_historical_chart,
                inputs=period_selector,
                outputs=chart_output
            )
        
        # Tab 4: Gamification
        with gr.Tab("🏆 Progress & Rewards"):
            gamification_display = gr.Markdown(value=get_gamification_data())
            refresh_gamification_btn = gr.Button("🔄 Refresh Stats")
            refresh_gamification_btn.click(fn=get_gamification_data, outputs=gamification_display)
        
        # Tab 5: Settings & Info
        with gr.Tab("⚙️ Settings & Info"):
            gr.Markdown("""
            ## About Plant Intelligence Hub
            
            This application provides:
            - 🌡️ Real-time environmental monitoring
            - 🤖 AI-powered plant health analysis
            - 📊 Historical data tracking and visualization
            - 🏆 Gamification to encourage regular plant care
            
            ### Features:
            - **Sensor Monitoring:** Track temperature, humidity, soil moisture, and light levels
            - **Image Analysis:** Upload plant photos for disease detection and health assessment
            - **Trend Analysis:** View historical data to identify patterns
            - **Achievements:** Complete daily missions and unlock rewards
            - **Leaderboard:** Compete with other plant enthusiasts
            
            ### Technology Stack:
            - Python 3.x
            - Gradio for UI
            - Plotly for data visualization
            - PIL for image processing
            
            ---
            
            **Converted from React/TypeScript to Python for Google Colab**
            
            Original project: Plant Intelligence Hub (React)
            """)
    
    # Footer
    gr.Markdown("""
    ---
    💚 Keep your plants healthy and happy! | 🤖 Powered by AI
    """)

# Launch the app
print("🚀 Launching Plant Intelligence Hub...")
demo.launch(share=True, debug=True)
```
