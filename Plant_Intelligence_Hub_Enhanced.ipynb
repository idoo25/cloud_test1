## Step 3: Data Generation Functions

```python
def generate_sensor_data():
    """Generate current sensor readings"""
    return {
        'temperature': {
            'value': round(20 + random.uniform(0, 10), 1),
            'unit': '°C',
            'status': 'good',
            'min': 18,
            'max': 30,
            'icon': '🌡️',
            'color': '#ef4444'
        },
        'humidity': {
            'value': round(50 + random.uniform(0, 30), 1),
            'unit': '%',
            'status': 'good',
            'min': 50,
            'max': 80,
            'icon': '💧',
            'color': '#3b82f6'
        },
        'soil_moisture': {
            'value': round(30 + random.uniform(0, 30), 1),
            'unit': '%',
            'status': 'warning',
            'min': 30,
            'max': 60,
            'icon': '🌱',
            'color': '#06b6d4'
        },
        'light_intensity': {
            'value': int(200 + random.uniform(0, 800)),
            'unit': 'lux',
            'status': 'good',
            'min': 600,
            'max': 1000,
            'icon': '☀️',
            'color': '#f59e0b'
        }
    }

def generate_historical_data(days=7):
    """Generate historical sensor data"""
    hours = days * 24
    timestamps = [datetime.now() - timedelta(hours=i) for i in range(hours, 0, -1)]
    
    data = {
        'timestamp': timestamps,
        'temperature': [20 + random.uniform(0, 10) for _ in range(hours)],
        'humidity': [50 + random.uniform(0, 30) for _ in range(hours)],
        'soil_moisture': [30 + random.uniform(0, 30) for _ in range(hours)],
        'light_intensity': [200 + random.uniform(0, 800) for _ in range(hours)]
    }
    
    return pd.DataFrame(data)
```

## Step 5: Create Historical Chart

```python
def create_historical_chart(period='week'):
    """Create dark-themed interactive chart"""
    days_map = {'week': 7, 'month': 30, 'quarter': 90}
    days = days_map.get(period, 7)
    
    df = generate_historical_data(days)
    
    fig = make_subplots(
        rows=2, cols=2,
        subplot_titles=(
            '🌡️ Temperature (°C)',
            '💧 Humidity (%)',
            '🌱 Soil Moisture (%)',
            '☀️ Light Intensity (lux)'
        ),
        vertical_spacing=0.15,
        horizontal_spacing=0.12
    )
    
    # Add traces
    fig.add_trace(
        go.Scatter(
            x=df['timestamp'],
            y=df['temperature'],
            name='Temperature',
            line=dict(color='#ef4444', width=3),
            fill='tozeroy',
            fillcolor='rgba(239, 68, 68, 0.1)'
        ),
        row=1, col=1
    )
    
    fig.add_trace(
        go.Scatter(
            x=df['timestamp'],
            y=df['humidity'],
            name='Humidity',
            line=dict(color='#3b82f6', width=3),
            fill='tozeroy',
            fillcolor='rgba(59, 130, 246, 0.1)'
        ),
        row=1, col=2
    )
    
    fig.add_trace(
        go.Scatter(
            x=df['timestamp'],
            y=df['soil_moisture'],
            name='Soil Moisture',
            line=dict(color='#06b6d4', width=3),
            fill='tozeroy',
            fillcolor='rgba(6, 182, 212, 0.1)'
        ),
        row=2, col=1
    )
    
    fig.add_trace(
        go.Scatter(
            x=df['timestamp'],
            y=df['light_intensity'],
            name='Light Intensity',
            line=dict(color='#f59e0b', width=3),
            fill='tozeroy',
            fillcolor='rgba(245, 158, 11, 0.1)'
        ),
        row=2, col=2
    )
    
    # Update layout with dark theme
    fig.update_layout(
        height=800,
        showlegend=False,
        title_text=f"📈 Historical Trends - Last {period.capitalize()}",
        title_font=dict(size=24, color='white', family='Arial'),
        template='plotly_dark',
        paper_bgcolor='rgba(10, 10, 15, 0.95)',
        plot_bgcolor='rgba(26, 26, 46, 0.5)',
        margin=dict(t=100, l=80, r=80, b=80),
        font=dict(color='white')
    )
    
    # Update axes
    fig.update_xaxes(
        showgrid=True,
        gridwidth=1,
        gridcolor='rgba(168, 85, 247, 0.1)',
        showline=True,
        linewidth=2,
        linecolor='rgba(168, 85, 247, 0.3)'
    )
    
    fig.update_yaxes(
        showgrid=True,
        gridwidth=1,
        gridcolor='rgba(168, 85, 247, 0.1)',
        showline=True,
        linewidth=2,
        linecolor='rgba(168, 85, 247, 0.3)'
    )
    
    # Update subplot titles
    for annotation in fig['layout']['annotations']:
        annotation['font'] = dict(size=16, color='white')
    
    return fig
```
